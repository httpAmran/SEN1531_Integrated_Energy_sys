```python
import pandas as pd

print("Termination:", model.results.attrs.get("termination_condition"))

# Energy totals (MWh over model horizon)
demand_served = model.results.flow_in.sel(techs="demand_power", carriers="power").sum().item()
curtailed = model.results.flow_in.sel(techs="curtailment", carriers="power").sum().item()
load_shed = model.results.flow_out.sel(techs="load_shedding", carriers="power").sum().item()
wind_gen = model.results.flow_out.sel(techs="wind_offshore", carriers="power").sum().item()

# Available wind energy from profile * installed cap
wind_avail = (
    model.inputs.source_use_equals.sel(techs="wind_offshore", nodes="NLD")
    * model.inputs.flow_cap.sel(techs="wind_offshore", nodes="NLD")
).sum().item()

curtailment_rate = (curtailed / wind_avail * 100) if wind_avail > 0 else 0

# Shadow price stats


sp = model.results.shadow_price_system_balance.to_series()
price_stats = sp.describe(percentiles=[0.5, 0.95])[["mean", "50%", "95%", "max"]]

kpis = pd.Series({
    "Demand served (MWh)": demand_served,
    "Wind available (MWh)": wind_avail,
    "Wind generated (MWh)": wind_gen,
    "Curtailment (MWh)": curtailed,
    "Curtailment rate (%)": curtailment_rate,
    "Load shedding (MWh)": load_shed,
})

print(kpis)
print("\nSystem balance price stats:")
print(price_stats)

```

```python
import pandas as pd

m = calliope.read_netcdf(f"results/{scenario_name}_results.nc")

# Common KPIs
demand_served = m.results.flow_in.sel(techs="demand_power", carriers="power").sum().item()
curtailed = m.results.flow_in.sel(techs="curtailment", carriers="power").sum().item()
load_shed = m.results.flow_out.sel(techs="load_shedding", carriers="power").sum().item()
wind_gen = m.results.flow_out.sel(techs="wind_offshore", carriers="power").sum().item()

if "source_use_equals" in m.inputs.data_vars:
    wind_avail = (
        m.inputs.source_use_equals.sel(techs="wind_offshore", nodes="NLD")
        * m.inputs.flow_cap.sel(techs="wind_offshore", nodes="NLD")
    ).sum().item()
else:
    wind_avail = (
        m.inputs.source_use_max.sel(techs="wind_offshore", nodes="NLD")
        * m.inputs.flow_cap.sel(techs="wind_offshore", nodes="NLD")
    ).sum().item()

curtailment_rate = (curtailed / wind_avail * 100) if wind_avail > 0 else 0
sp = m.results.shadow_price_system_balance.sel(nodes="NLD", carriers="power").to_series().dropna()
price_stats = sp.describe(percentiles=[0.5, 0.95])[["mean", "50%", "95%", "max"]]
pd.set_option("display.float_format", "{:,.2f}".format)

# Battery-specific KPIs
battery_charge = m.results.flow_in.sel(techs="battery_storage", carriers="power").sum().item()
battery_discharge = m.results.flow_out.sel(techs="battery_storage", carriers="power").sum().item()
battery_losses = battery_charge - battery_discharge
battery_round_trip = (battery_discharge / battery_charge * 100) if battery_charge > 0 else 0

kpis_battery = pd.Series({
    "Demand served (MWh)": demand_served,
    "Wind available (MWh)": wind_avail,
    "Wind generated (MWh)": wind_gen,
    "Curtailment (MWh)": curtailed,
    "Curtailment rate (%)": curtailment_rate,
    "Load shedding (MWh)": load_shed,
    "Battery charge (MWh)": battery_charge,
    "Battery discharge (MWh)": battery_discharge,
    "Battery losses (MWh)": battery_losses,
    "Battery round-trip (%)": battery_round_trip,
})

print(kpis_battery)
print("\nPower system balance price stats:")
print(price_stats)
```

```python
m_h2 = calliope.read_netcdf(f"results/{scenario_name_h2}_results.nc")

demand_served_h2 = m_h2.results.flow_in.sel(techs="demand_power", carriers="power").sum().item()
curtailed_h2 = m_h2.results.flow_in.sel(techs="curtailment", carriers="power").sum().item()
load_shed_h2 = m_h2.results.flow_out.sel(techs="load_shedding", carriers="power").sum().item()
wind_gen_h2 = m_h2.results.flow_out.sel(techs="wind_offshore", carriers="power").sum().item()

if "source_use_equals" in m_h2.inputs.data_vars:
    wind_avail_h2 = (m_h2.inputs.source_use_equals.sel(techs="wind_offshore", nodes="NLD") * m_h2.inputs.flow_cap.sel(techs="wind_offshore", nodes="NLD")).sum().item()
else:
    wind_avail_h2 = (m_h2.inputs.source_use_max.sel(techs="wind_offshore", nodes="NLD") * m_h2.inputs.flow_cap.sel(techs="wind_offshore", nodes="NLD")).sum().item()

curtailment_rate_h2 = (curtailed_h2 / wind_avail_h2 * 100) if wind_avail_h2 > 0 else 0
sp_h2 = m_h2.results.shadow_price_system_balance.sel(nodes="NLD", carriers="power").to_series().dropna()
price_stats_h2 = sp_h2.describe(percentiles=[0.5, 0.95])[["mean", "50%", "95%", "max"]]

electrolyser_in = m_h2.results.flow_in.sel(techs="electrolyser", carriers="power").sum().item()
fuel_cell_out = m_h2.results.flow_out.sel(techs="fuel_cell", carriers="power").sum().item() if "fuel_cell" in m_h2.results.coords["techs"].values else 0.0
h2_round_trip = (fuel_cell_out / electrolyser_in * 100) if electrolyser_in > 0 else 0

kpis_hydrogen = pd.Series({
    "Demand served (MWh)": demand_served_h2,
    "Wind available (MWh)": wind_avail_h2,
    "Wind generated (MWh)": wind_gen_h2,
    "Curtailment (MWh)": curtailed_h2,
    "Curtailment rate (%)": curtailment_rate_h2,
    "Load shedding (MWh)": load_shed_h2,
    "Electrolyser input (MWh)": electrolyser_in,
    "Fuel cell output (MWh)": fuel_cell_out,
    "Power-to-power via H2 (%)": h2_round_trip,
})

print(kpis_hydrogen)
print("\nPower system balance price stats:")
print(price_stats_h2)
```

## 6. Compare Scenarios

Build one comparison table from saved NetCDF results.


```python
import numpy as np
import pandas as pd
from pathlib import Path

def _wind_available(model):
    if "source_use_equals" in model.inputs.data_vars:
        profile = model.inputs.source_use_equals.sel(techs="wind_offshore", nodes="NLD")
    else:
        profile = model.inputs.source_use_max.sel(techs="wind_offshore", nodes="NLD")
    cap = model.inputs.flow_cap.sel(techs="wind_offshore", nodes="NLD")
    return (profile * cap).sum().item()

def compute_kpis(nc_path, scenario_type):
    m = calliope.read_netcdf(nc_path)
    demand_served = m.results.flow_in.sel(techs="demand_power", carriers="power").sum().item()
    curtailed = m.results.flow_in.sel(techs="curtailment", carriers="power").sum().item()
    load_shed = m.results.flow_out.sel(techs="load_shedding", carriers="power").sum().item()
    wind_gen = m.results.flow_out.sel(techs="wind_offshore", carriers="power").sum().item()
    wind_avail = _wind_available(m)
    curtailment_rate = (curtailed / wind_avail * 100) if wind_avail > 0 else 0

    if "shadow_price_system_balance" in m.results.data_vars:
        sp = m.results.shadow_price_system_balance.sel(nodes="NLD", carriers="power").to_series().dropna()
        price_mean = sp.mean() if len(sp) else np.nan
        price_p95 = sp.quantile(0.95) if len(sp) else np.nan
        price_max = sp.max() if len(sp) else np.nan
    else:
        price_mean, price_p95, price_max = np.nan, np.nan, np.nan

    out = {
        "Demand served (MWh)": demand_served,
        "Wind available (MWh)": wind_avail,
        "Wind generated (MWh)": wind_gen,
        "Curtailment (MWh)": curtailed,
        "Curtailment rate (%)": curtailment_rate,
        "Load shedding (MWh)": load_shed,
        "Price mean (kEUR/MWh)": price_mean,
        "Price p95 (kEUR/MWh)": price_p95,
        "Price max (kEUR/MWh)": price_max,
    }

    if scenario_type == "battery":
        ch = m.results.flow_in.sel(techs="battery_storage", carriers="power").sum().item()
        dis = m.results.flow_out.sel(techs="battery_storage", carriers="power").sum().item()
        out["Battery discharge (MWh)"] = dis
        out["Battery round-trip (%)"] = (dis / ch * 100) if ch > 0 else np.nan
    elif scenario_type == "hydrogen":
        el = m.results.flow_in.sel(techs="electrolyser", carriers="power").sum().item()
        fc = m.results.flow_out.sel(techs="fuel_cell", carriers="power").sum().item() if "fuel_cell" in m.results.coords["techs"].values else 0.0
        out["Electrolyser input (MWh)"] = el
        out["Fuel cell output (MWh)"] = fc
        out["Power-to-power via H2 (%)"] = (fc / el * 100) if el > 0 else np.nan
    elif scenario_type == "big_user":
        bu = m.results.flow_in.sel(techs="demand_big_user", carriers="power").sum().item()
        out["Big user served (MWh)"] = bu
        out["Big user revenue equivalent (kEUR)"] = bu * 0.02

    return out

scenario_meta = {
    "surplus_to_battery": "battery",
    "surplus_to_hydrogen": "hydrogen",
    "surplus_to_big_user": "big_user",
    "surplus_to_big_user_two_nodes": "big_user",
}

rows = {}
for sc, kind in scenario_meta.items():
    nc = Path("results") / f"{sc}_results.nc"
    if not nc.exists():
        print(f"Skipping {sc}: {nc} not found. Run that scenario cell first.")
        continue
    rows[sc] = compute_kpis(str(nc), kind)

if not rows:
    raise RuntimeError("No scenario result files found. Run scenario solve cells first.")

kpi_compare = pd.DataFrame(rows).T
kpi_compare
```

## 8. Investment Inputs (Clear Output)
This section prints one clear scenario table for your next investment stage.

Included from Calliope:
- `incidental_opex_kEUR` (variable operating cost)
- electricity prices from `shadow_price_system_balance`
- electricity revenue (`flow_out * price`)
- `capex_proxy_kEUR` from fixed capacities and capex inputs

Note: fixed annual O&M is not defined in this model (`cost_om_annual` absent), so it is shown as `NaN`.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import calliope

RESULTS_DIR = Path('results')
PRICE_NODE = 'NLD'
PRICE_CARRIER = 'power'
COST_CLASS = 'monetary'
REVENUE_TECHS = ['wind_offshore', 'battery_storage', 'fuel_cell', 'nld_to_big_user_line']
EXCLUDE_SCENARIOS = {'carbontax', 'carbon_tax', 'less_generation','more_vres', 'surplus_to_big_user', 'baseline'}# Exclude scenarios that are not relevant for investment input summary

def _select_pref_or_first(da, dim, preferred):
    if dim not in da.dims:
        return da
    vals = da.coords[dim].values.tolist()
    return da.sel({dim: preferred if preferred in vals else vals[0]})

def get_price_series(m):
    if 'shadow_price_system_balance' not in m.results.data_vars:
        return pd.Series(dtype=float)
    p = m.results.shadow_price_system_balance
    p = _select_pref_or_first(p, 'nodes', PRICE_NODE)
    p = _select_pref_or_first(p, 'carriers', PRICE_CARRIER)
    return p.to_series().dropna()


def get_actual_sold_electricity(m):
    """Delivered electricity to end-use power demand sinks (no generation/storage double count)."""
    if 'flow_in' not in m.results.data_vars:
        return np.nan
    flow_in = m.results.flow_in
    if 'techs' not in flow_in.dims:
        return np.nan

    sold = 0.0
    for tech in flow_in.coords['techs'].values.tolist():
        if not str(tech).startswith('demand_'):
            continue

        fi = flow_in.sel(techs=tech)

        # Keep only electricity demand sinks; skip hydrogen/other carriers
        if 'carriers' in fi.dims:
            carriers = fi.coords['carriers'].values.tolist()
            if PRICE_CARRIER not in carriers:
                continue
            fi = fi.sel(carriers=PRICE_CARRIER)

        sold += float(fi.fillna(0).sum().item())

    return sold

def get_total_lcoe(m):
    if 'total_levelised_cost' not in m.results.data_vars:
        return np.nan
    l = m.results.total_levelised_cost
    l = _select_pref_or_first(l, 'costs', COST_CLASS)
    l = _select_pref_or_first(l, 'carriers', PRICE_CARRIER)
    return float(l.item())

rows = []
price_series_by_scenario = {}

for nc in sorted(RESULTS_DIR.glob('*.nc')):
    scenario = nc.stem.replace('_results', '')
    if scenario.lower() in EXCLUDE_SCENARIOS:
        continue

    m = calliope.read_netcdf(str(nc))

    price_s = get_price_series(m)
    price_series_by_scenario[scenario] = price_s

    incidental_opex = (
        float(m.results.cost_operation_variable.fillna(0).sum().item())
        if 'cost_operation_variable' in m.results.data_vars else np.nan
    )

    revenue = 0.0
    sold_mwh = 0.0
    actual_sold_mwh = get_actual_sold_electricity(m)
    techs_available = set(m.results.coords['techs'].values.tolist())

    for tech in REVENUE_TECHS:
        if tech not in techs_available:
            continue
        f = m.results.flow_out.sel(techs=tech)
        f = _select_pref_or_first(f, 'nodes', PRICE_NODE)
        f = _select_pref_or_first(f, 'carriers', PRICE_CARRIER)

        flow_s = f.to_series().fillna(0.0)
        sold_mwh += float(flow_s.sum())

        if not price_s.empty:
            merged = pd.concat([flow_s.rename('flow'), price_s.rename('price')], axis=1).fillna(0.0)
            revenue += float((merged['flow'] * merged['price']).sum())

    capex_proxy = 0.0
    if 'flow_cap' in m.inputs.data_vars and 'cost_flow_cap' in m.inputs.data_vars:
        capex_proxy += float((m.inputs.flow_cap.fillna(0) * m.inputs.cost_flow_cap.fillna(0)).sum().item())
    if 'storage_cap' in m.inputs.data_vars and 'cost_storage_cap' in m.inputs.data_vars:
        capex_proxy += float((m.inputs.storage_cap.fillna(0) * m.inputs.cost_storage_cap.fillna(0)).sum().item())

    rows.append({
        'scenario': scenario,
        'incidental_opex_kEUR': incidental_opex,
        # 'fixed_annual_om_kEUR': np.nan,
        'shadow_price_avg_kEUR_per_MWh': float(price_s.mean()) if not price_s.empty else np.nan,
        'shadow_price_p50_kEUR_per_MWh': float(price_s.quantile(0.5)) if not price_s.empty else np.nan,
        'lcoe_total_kEUR_per_MWh': get_total_lcoe(m),
        'electricity_sold_MWh': sold_mwh,
        'actual_sold_electricity_MWh': actual_sold_mwh,
        'electricity_revenue_kEUR': revenue,
        'capex_proxy_kEUR': capex_proxy,
    })

investment_summary_df = pd.DataFrame(rows).sort_values('scenario').reset_index(drop=True)

print('Shadow-price source: results.shadow_price_system_balance (NLD, power)')
print('LCOE source: results.total_levelised_cost (monetary, power)')

important_cols = [
    'scenario',
    'incidental_opex_kEUR',
    'shadow_price_avg_kEUR_per_MWh',
    'shadow_price_p50_kEUR_per_MWh',
    'lcoe_total_kEUR_per_MWh',
    'electricity_sold_MWh',
    'actual_sold_electricity_MWh',
    'electricity_revenue_kEUR',
    'capex_proxy_kEUR',
]

investment_summary_display = investment_summary_df.copy()
# Keep row sort stable across runs
investment_summary_display = investment_summary_display.sort_values('scenario').reset_index(drop=True)

display(investment_summary_display[important_cols].round(6))

# Totals / aggregate checks for investment review
totals = investment_summary_display[important_cols[1:]].sum(numeric_only=True)
print('\nTotals across included scenarios:')
# print(totals.round(8))

investment_summary_df.to_csv(RESULTS_DIR / 'investment_inputs_summary.csv', index=False)
```

```python
# 11. Clean Scenario Business Table
# Presentation-friendly table with only positive values:
# gross OPEX, CAPEX proxy, and pathway product sold energy per scenario.

from pathlib import Path
import pandas as pd
import numpy as np
import calliope

RESULTS_DIR = Path('results')

SCENARIO_SPECS = [
    {
        'scenario': 'default',
        'sold_energy_def': 'electricity_to_demand_power_MWh',
        'sold_energy': {'var': 'flow_in', 'tech': 'demand_power', 'carrier': 'power'},
    },
    {
        'scenario': 'surplus_to_battery',
        'sold_energy_def': 'battery_discharge_MWh',
        'sold_energy': {'var': 'flow_out', 'tech': 'battery_storage', 'carrier': 'power'},
    },
    {
        'scenario': 'surplus_to_hydrogen',
        'sold_energy_def': 'hydrogen_sold_MWh_h2',
        'sold_energy': {'var': 'flow_in', 'tech': 'demand_h2_offtake', 'carrier': 'hydrogen'},
    },
    {
        'scenario': 'surplus_to_big_user_two_nodes',
        'sold_energy_def': 'electricity_to_big_user_MWh',
        'sold_energy': {'var': 'flow_in', 'tech': 'demand_big_user', 'carrier': 'power'},
    },
]


def _sum_da(da):
    return float(da.fillna(0).sum().item())


def _sum_flow(m, var_name, tech, carrier):
    if var_name not in m.results.data_vars:
        return 0.0
    da = m.results[var_name]
    if 'techs' not in da.dims or tech not in da.coords['techs'].values.tolist():
        return 0.0
    da = da.sel(techs=tech)
    if 'carriers' in da.dims:
        carriers = da.coords['carriers'].values.tolist()
        if carrier not in carriers:
            return 0.0
        da = da.sel(carriers=carrier)
    return _sum_da(da)


def _scenario_opex_gross(m):
    """Gross variable operating cost only (no negative revenue terms)."""
    if 'cost_operation_variable' not in m.results.data_vars:
        return np.nan
    vals = m.results.cost_operation_variable.fillna(0).to_series()
    return float(vals[vals > 0].sum())


def _scenario_capex_proxy(m):
    capex = 0.0
    if 'flow_cap' in m.inputs.data_vars and 'cost_flow_cap' in m.inputs.data_vars:
        capex += _sum_da(m.inputs.flow_cap * m.inputs.cost_flow_cap)
    if 'storage_cap' in m.inputs.data_vars and 'cost_storage_cap' in m.inputs.data_vars:
        capex += _sum_da(m.inputs.storage_cap * m.inputs.cost_storage_cap)
    return capex


rows = []
for spec in SCENARIO_SPECS:
    nc = RESULTS_DIR / f"{spec['scenario']}_results.nc"
    if not nc.exists():
        continue

    m = calliope.read_netcdf(str(nc))

    sold = _sum_flow(
        m,
        spec['sold_energy']['var'],
        spec['sold_energy']['tech'],
        spec['sold_energy']['carrier'],
    )

    rows.append({
        'scenario': spec['scenario'],
        'opex_gross_kEUR': _scenario_opex_gross(m),
        'capex_proxy_kEUR': _scenario_capex_proxy(m),
        'sold_energy_MWh': sold,
        'sold_energy_definition': spec['sold_energy_def'],
    })

clean_business_df = pd.DataFrame(rows)

display(clean_business_df.round(6))
clean_business_df.to_csv(RESULTS_DIR / 'scenario_business_clean.csv', index=False)
print('Saved:', RESULTS_DIR / 'scenario_business_clean.csv')

```

```python
# 12. Scenario Comparison Table — Full KPI Summary (Slide-ready)

from pathlib import Path
import numpy as np
import pandas as pd
import calliope, warnings
warnings.filterwarnings("ignore")

RESULTS_DIR = Path("results")

SCENARIOS = {
    "baseline":                     "Baseline",
    "surplus_to_battery":           "Battery",
    "surplus_to_hydrogen":          "Hydrogen",
    "surplus_to_big_user":          "Big User (1 node)",
    "surplus_to_big_user_two_nodes":"Big User (2 nodes)",
}

# ── helpers ───────────────────────────────────────────────────────────────────
def _wind_avail(m):
    p = m.inputs.source_use_equals if "source_use_equals" in m.inputs.data_vars \
        else m.inputs.source_use_max
    cap = m.inputs.flow_cap.sel(techs="wind_offshore", nodes="NLD")
    return float((p.sel(techs="wind_offshore", nodes="NLD") * cap).sum().item())

def _sum(m, var, tech, carrier):
    if var not in m.results.data_vars: return 0.0
    da = m.results[var]
    if "techs" not in da.dims or tech not in da.coords["techs"].values.tolist(): return 0.0
    da = da.sel(techs=tech)
    if "carriers" in da.dims:
        if carrier not in da.coords["carriers"].values.tolist(): return 0.0
        da = da.sel(carriers=carrier)
    return float(da.fillna(0).sum().item())

def _total_cost(m):
    if "cost" not in m.results.data_vars: return np.nan
    return float(m.results.cost.sel(costs="monetary").fillna(0).sum().item())

# ── collect rows ──────────────────────────────────────────────────────────────
raw = {}
for sc, label in SCENARIOS.items():
    nc = RESULTS_DIR / f"{sc}_results.nc"
    if not nc.exists():
        print(f"  Missing {nc.name} — skipping"); continue
    m = calliope.read_netcdf(str(nc))

    wind_avail    = _wind_avail(m)
    wind_gen      = _sum(m, "flow_out", "wind_offshore",  "power")
    curtailed     = _sum(m, "flow_in",  "curtailment",    "power")
    load_shed     = _sum(m, "flow_out", "load_shedding",  "power")
    demand_served = _sum(m, "flow_in",  "demand_power",   "power")
    ccgt_gen      = _sum(m, "flow_out", "ccgt",           "power")
    total_cost    = _total_cost(m)
    curt_rate     = curtailed / wind_avail * 100 if wind_avail > 0 else 0.0

    # Shadow / electricity price
    sp = m.results.shadow_price_system_balance.sel(nodes="NLD", carriers="power").to_series().dropna()
    neg_price_hours  = int((sp < 0).sum())
    mean_price_eur   = sp.mean() * 1000          # kEUR/MWh -> EUR/MWh
    p95_price_eur    = sp.quantile(0.95) * 1000

    # Pathway-specific
    bat_discharge = _sum(m, "flow_out", "battery_storage", "power")
    elec_input    = _sum(m, "flow_in",  "electrolyser",    "power")
    big_user      = _sum(m, "flow_in",  "demand_big_user", "power")

    raw[sc] = dict(
        label=label,
        total_cost=total_cost,
        wind_avail=wind_avail,
        wind_gen=wind_gen,
        curtailed=curtailed,
        curt_rate=curt_rate,
        load_shed=load_shed,
        demand_served=demand_served,
        ccgt_gen=ccgt_gen,
        neg_price_hours=neg_price_hours,
        mean_price_eur=mean_price_eur,
        p95_price_eur=p95_price_eur,
        bat_discharge=bat_discharge,
        elec_input=elec_input,
        big_user=big_user,
    )

# ── curtailment reduction vs baseline ─────────────────────────────────────────
baseline_curt = raw["baseline"]["curtailed"] if "baseline" in raw else np.nan

rows = []
for sc, d in raw.items():
    curt_red_abs = baseline_curt - d["curtailed"]
    curt_red_pct = curt_red_abs / baseline_curt * 100 if baseline_curt > 0 else 0.0

    rows.append({
        "Scenario":                        d["label"],
        "Total cost (M EUR)":              round(d["total_cost"] / 1e3, 1),
        "Wind generated (TWh)":            round(d["wind_gen"] / 1e6, 2),
        "Curtailment (TWh)":               round(d["curtailed"] / 1e6, 2),
        "Curtailment rate (%)":            round(d["curt_rate"], 1),
        "Curt. reduction vs baseline (TWh)": round(curt_red_abs / 1e6, 2),
        "Curt. reduction vs baseline (%)":   round(curt_red_pct, 1),
        "CCGT dispatch (TWh)":             round(d["ccgt_gen"] / 1e6, 2),
        "Demand served (TWh)":             round(d["demand_served"] / 1e6, 2),
        "Load shedding (MWh)":             round(d["load_shed"], 0),
        "Neg. price hours":                d["neg_price_hours"],
        "Mean price (EUR/MWh)":            round(d["mean_price_eur"], 2),
        "p95 price (EUR/MWh)":             round(d["p95_price_eur"], 2),
        "Battery discharge (TWh)":         round(d["bat_discharge"] / 1e6, 2) if d["bat_discharge"] > 0 else "—",
        "Electrolyser input (TWh)":        round(d["elec_input"] / 1e6, 2)   if d["elec_input"]   > 0 else "—",
        "Big user served (TWh)":           round(d["big_user"] / 1e6, 2)     if d["big_user"]     > 0 else "—",
    })

df = pd.DataFrame(rows).set_index("Scenario")

# ── styling ────────────────────────────────────────────────────────────────────
LOWER_IS_BETTER  = [
    "Total cost (M EUR)", "Curtailment (TWh)", "Curtailment rate (%)",
    "CCGT dispatch (TWh)", "Load shedding (MWh)", "Neg. price hours",
]
HIGHER_IS_BETTER = [
    "Curt. reduction vs baseline (TWh)", "Curt. reduction vs baseline (%)",
]

def _highlight(s, lower_is_better=True):
    nums = pd.to_numeric(s, errors="coerce")
    if nums.isna().all(): return ["" for _ in s]
    best = nums.min() if lower_is_better else nums.max()
    return ["background-color: #d4edda; font-weight: bold" if v == best else "" for v in nums]

styled = df.style
for col in LOWER_IS_BETTER:
    if col in df.columns:
        styled = styled.apply(_highlight, lower_is_better=True,  subset=[col])
for col in HIGHER_IS_BETTER:
    if col in df.columns:
        styled = styled.apply(_highlight, lower_is_better=False, subset=[col])

styled = (styled
    .set_caption("Scenario Comparison — Full KPI Summary  |  green = best in column")
    .format(na_rep="—")
    .set_table_styles([{"selector": "caption",
                        "props": [("font-size", "14px"), ("font-weight", "bold"),
                                  ("text-align", "left"), ("padding-bottom", "6px")]}])
)

display(styled)

df.to_csv(RESULTS_DIR / "scenario_comparison_table.csv")
print("Saved: results/scenario_comparison_table.csv")
```
